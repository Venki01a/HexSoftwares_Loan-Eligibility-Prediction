# EDA

# Remove leading spaces from column names


```python
data.rename(columns=lambda x: x.strip(), inplace=True)
```

# Display the updated columns

```python
data.columns
```

# Drop Loan id because it is not important for prediction

```python
data.drop('loan_id', axis=1, inplace=True)
```

# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
le
```

```python
# Apply label encoding to the 'education' column
data['education'] = le.fit_transform(data['education'])
```

```python
# Apply label encoding to 'self_employed' column
data['self_employed'] = le.fit_transform(data['self_employed'])
```

```python
# Apply label encoding to 'loan status' column
data['loan_status'] = le.fit_transform(data['loan_status'])
```

# Display the encoded colums

```python
data['education'].value_counts()
```

```python
data['self_employed'].value_counts()
```

```python
data['loan_status'].value_counts()
```

# define x and y

```python
x=data.drop('loan_status', axis=1)
y=data['loan_status']
```

# Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
sc
```

# Select only the numerical columns for scaling (excluding 'loan_status')

```python
numerical_columns = [ 'income_annum', 'loan_amount', 'loan_term', 'cibil_score',
                      'residential_assets_value', 'commercial_assets_value', 'luxury_assets_value',
                      'bank_asset_value']
```

# Apply scaling to the numerical columns

```python
x[numerical_columns] = sc.fit_transform(x[numerical_columns])
```

# Display the scaled features of the variable x

```python
print("Features after scaling:")
x.head()
```

# Display the target variable y

```python
print("Target variable:")
y
```

# Train - Test Split

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

# Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression
lr_model = LogisticRegression()
lr_model
```

```python
lr_model.fit(x_train, y_train)
```

# Predictions by logistic regression model

```python
lr_model.predict(x_test)
```

# Accuracy From logistic Regression Model

```python
from sklearn.metrics import accuracy_score
lr_model_accuracy = accuracy_score(y_test, lr_model.predict(x_test))
lr_model_accuracy
```

# Classification report of logistic regression model

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, lr_model.predict(x_test)))
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier()
dt_model
```

```python
dt_model.fit(x_train, y_train)
```

# Prediction by Decision Tree

```python
dt_model.predict(x_test)
```

# Accuracy from Decision tree

```python
dt_model.accuracy = accuracy_score(y_test, dt_model.predict(x_test))
dt_model.accuracy
```

# Classification report of Decision Tree

```python
print(classification_report(y_test, dt_model.predict(x_test)))
```
